```python
def top5_classes_weighted_accuracy(csv_path):
    df = pd.read_csv(csv_path)
    
    # Get top 5 classes by accuracy
    top5_df = df.nlargest(5, 'Accuracy (%)')

    # Calculate weighted accuracy
    correct_preds = (top5_df['Accuracy (%)'] * top5_df['Total']).sum()
    total_samples = top5_df['Total'].sum()
    
    weighted_accuracy = correct_preds / total_samples
    
    return weighted_accuracy, top5_df

weighted_acc, top5_classes = top5_classes_weighted_accuracy("RGBVIDEOTrained/rgb_per_class_accuracy.csv")

print("Weighted Accuracy for Top-5 Classes:", weighted_acc)
print(top5_classes)
```

```python
import os
import pandas as pd

def top5_classes_weighted_accuracyForPointTracks(csv_path):
    df = pd.read_csv(csv_path)
    
    # Ensure correct columns exist
    if 'Accuracy' not in df.columns or 'Total_Samples' not in df.columns:
        raise ValueError(f"Missing required columns in {csv_path}")

    # Get top 5 by Accuracy
    top5_df = df.nlargest(5, 'Accuracy')

    # Weighted accuracy
    correct_preds = (top5_df['Accuracy'] * top5_df['Total_Samples']).sum()
    total_samples = top5_df['Total_Samples'].sum()
    
    weighted_accuracy = correct_preds / total_samples
    return weighted_accuracy, top5_df

# === Set your directory path here ===
root_dir = "/Users/mrinalraj/Downloads/WebDownload/Driving48/TrainingPointTemporalGRU/TrainingAgain"
# /Users/mrinalraj/Downloads/WebDownload/Driving48/TrainingPointTemporalGRU/TrainingAgain
# === Iterate over Epoch_* folders ===
for folder_name in sorted(os.listdir(root_dir)):
    folder_path = os.path.join(root_dir, folder_name)
    csv_path = os.path.join(folder_path, "per_class_accuracy.csv")

    if os.path.isdir(folder_path) and folder_name.startswith("Epoch_") and os.path.exists(csv_path):
        try:
            weighted_acc, top5_classes = top5_classes_weighted_accuracyForPointTracks(csv_path)
            print(f"\nEpoch: {folder_name}")
            print("Weighted Accuracy for Top-5 Classes:", round(weighted_acc, 4))
            print(top5_classes[['Class', 'Accuracy', 'Total_Samples']])
        except Exception as e:
            print(f"❌ Error processing {folder_name}: {e}")
```
