```python
import torch
import pandas as pd
import numpy as np
import os
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

# Make sure Dataloader.py is in the same directory
from TrainONNew.Dataloader import MotionDataset
```

```python

# --- Configuration ---
PT_FOLDER = "videosTensors1000"
OUTPUT_VIDEO_FILE = "augmentation_visualization.mp4"
MAX_POINTS = 1000 # Keep this low for a clear visualization
NUM_AUGMENTATIONS = 3 # How many augmented versions to show
SEQ_LENGTH = 120 # Number of frames in the dummy video


def main():
    TRAIN_DF_PATH = "/Users/mrinalraj/Downloads/WebDownload/Driving48/df_exsists.csv"
    train_df = pd.read_csv(TRAIN_DF_PATH)
    dataset = MotionDataset(
        df=train_df, pt_folder=PT_FOLDER, training=True, max_points=MAX_POINTS, p_aug=1.0
    )

    # Get one processed sample. The features are (x, y, visibility, vx, vy)
    original_tensor, _, _ = dataset[5]
    
    print("⚙️  Generating augmented versions...")
    augmented_tensors = [dataset._augment_motion(original_tensor.clone()) for _ in range(NUM_AUGMENTATIONS)]
    
    all_tensors = [original_tensor] + augmented_tensors
    titles = ["Original"] + [f"Augmented {i+1}" for i in range(NUM_AUGMENTATIONS)]
    
    # --- Set up the plot for a cleaner "video" style ---
    num_plots = len(all_tensors)
    fig, axes = plt.subplots(1, num_plots, figsize=(5 * num_plots, 5.5), dpi=120)
    fig.suptitle("Motion & Visibility Augmentation", fontsize=16, color='white')
    fig.patch.set_facecolor('black') # Dark background for the whole figure

    scatters = []
    for i, ax in enumerate(axes):
        ax.set_title(titles[i], color='white')
        ax.set_facecolor('#1a1a1a') # Dark plot background
        # Set fixed limits and remove ticks for a cleaner look
        ax.set_xlim(-0.2, 1.2)
        ax.set_ylim(-0.2, 1.2)
        ax.set_xticks([])
        ax.set_yticks([])
        ax.set_aspect('equal', 'box')
        # Initial scatter plot with a bright color
        scatters.append(ax.scatter([], [], s=25, c='#00FFFF')) # Bright Cyan

    frame_text = fig.text(0.5, 0.04, '', ha='center', fontsize=12, color='white')

    def update(frame_idx):
        """Update function for the animation, now using visibility for alpha."""
        artists = []
        for i, tensor in enumerate(all_tensors):
            coords = tensor[frame_idx, :, :2]     # Position (x, y)
            visibility = tensor[frame_idx, :, 2]  # Visibility (0 to 1)

            # Create an RGBA color array. Alpha channel is now controlled by visibility.
            colors = np.zeros((MAX_POINTS, 4))
            colors[:, :3] = [0, 1, 1]  # RGB for Cyan
            colors[:, 3] = visibility.numpy() # Set alpha (transparency) from visibility
            
            scatters[i].set_offsets(coords.numpy())
            scatters[i].set_facecolors(colors) # Update colors and transparency
            artists.append(scatters[i])
        
        frame_text.set_text(f'Frame: {frame_idx + 1} / {SEQ_LENGTH}')
        artists.append(frame_text)
        return artists

    print("📹 Creating animation... This may take a moment.")
    ani = FuncAnimation(fig, update, frames=SEQ_LENGTH, blit=True, interval=50)
    
    try:
        ani.save(OUTPUT_VIDEO_FILE, writer='ffmpeg', fps=20, savefig_kwargs={'facecolor': 'black'})
        print(f"🎉 Successfully saved visualization to: {OUTPUT_VIDEO_FILE}")
    except Exception as e:
        print(f"❌ Error saving animation: {e}")
        print("Please ensure 'ffmpeg' is installed and accessible in your system's PATH.")



if __name__ == '__main__':
    main()
```
