```python
import torch
import numpy as np
import cv2
import os

def visualize_tracking_from_pt(pt_path, save_path='output.mp4', visibility_threshold=0.5, canvas_size=(384, 512), fps=20):
    """
    Visualize tracking points from a .pt file containing 'pred_tracks' and 'pred_visibility'.
    
    Args:
        pt_path (str): Path to the .pt file
        save_path (str): Path to save the output video
        visibility_threshold (float): Threshold to show visible points
        canvas_size (tuple): (height, width) of the blank canvas
        fps (int): Frames per second for the output video
    """
    data = torch.load(pt_path)
    pred_tracks = data['pred_tracks'].cpu().numpy()       # [T, N, 2]
    pred_visibility = data['pred_visibility'].cpu().numpy()  # [T, N]

    T, N, _ = pred_tracks.shape
    H, W = canvas_size

    fourcc = cv2.VideoWriter_fourcc(*'mp4v')
    out = cv2.VideoWriter(save_path, fourcc, fps, (W, H))

    for t in range(T):
        frame = np.zeros((H, W, 3), dtype=np.uint8)

        for n in range(N):
            vis = pred_visibility[t, n]
            if vis > visibility_threshold:
                x, y = pred_tracks[t, n]
                x, y = int(x), int(y)
                if 0 <= x < W and 0 <= y < H:
                    cv2.circle(frame, (x, y), radius=2, color=(0, 255, 0), thickness=-1)

        out.write(frame)

    out.release()
    print(f"Saved visualization video to {save_path}")
```
