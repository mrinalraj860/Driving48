```python
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np

def plot_gif_frames(gif_path, num_frames=5):
    """
    Extracts evenly spaced frames from a GIF and plots them.

    Parameters:
    gif_path (str): Path to the GIF file
    num_frames (int): Number of frames to extract and display (default=5)
    """
    # Load GIF
    gif = Image.open(gif_path)
    
    # Get total number of frames
    total_frames = gif.n_frames
    
    # Choose equally spaced frame indices
    frame_indices = np.linspace(0, total_frames - 1, num_frames, dtype=int)
    
    # Plot frames
    fig, axes = plt.subplots(1, num_frames, figsize=(15, 3))
    for ax, idx in zip(axes, frame_indices):
        gif.seek(idx)
        ax.imshow(gif.convert("RGB"))
        ax.set_title(f"Frame {idx}/{total_frames-1}")
        ax.axis("off")
    
    plt.tight_layout()
    plt.savefig("temp.png")
    plt.show()

```
