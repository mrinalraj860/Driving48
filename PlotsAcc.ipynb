```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/Users/mrinalraj/Downloads/WebDownload/Driving48/TrainingVelocityTemporalCNNClassifier/training_metrics.csv")

epochs = df['Epoch']
train_acc = df['Train_Accuracy']
test_acc = df['Test_Accuracy']

```

```python
plt.figure(figsize=(10, 6))
plt.plot(epochs, train_acc, label='Train Accuracy', marker='o')
plt.plot(epochs, test_acc, label='Test Accuracy', marker='s')

plt.title('Train vs Test Accuracy over Epochs', fontsize=14)
plt.xlabel('Epoch', fontsize=12)
plt.ylabel('Accuracy', fontsize=12)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.savefig("train_test_accuracyTemporalCNNClassifier.png", dpi=300)  # high resolution for publications
plt.show()
```
